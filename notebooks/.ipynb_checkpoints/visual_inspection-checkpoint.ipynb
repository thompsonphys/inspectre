```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load data
traj_data_path = "../trajectory.dat"
lam, t, r, phi = np.loadtxt(traj_data_path).T
```

```python
fig, axs = plt.subplots(nrows=1, ncols=3, figsize=(20,5))
axs[0].plot(t,lam)
axs[0].set_ylabel(r'$\lambda$')
axs[1].plot(t,r)
axs[1].set_ylabel(r'$r$')
axs[2].plot(t,phi)
axs[2].set_ylabel(r'$\varphi$')

for ax in axs:
    ax.set_xlabel(r'$t$')
```
